```python
import pandas as pd
from astropy.io import fits

import numpy as np
import matplotlib.pyplot as plt
```

```python
# KOA_462
fits_file = fits.open('/Users/erin/Documents/Projects/KOA_Climate/data/KOA_462/NIRSPEC/extracted/fitstbl/flux/NS.20000325.45349_47_flux_tbl.fits.gz')

data_class = fits_file[1] # Get the [1]th element of fits_file
data = data_class.data # Get the data from data_class

a = data['wave (A)'] # Get the wavelength column
q = data['flux (cnts)'] # Get the flux column
a = a / 10000 # Convert the wavelength from Angstroms to microns
```

```python
# KOA_104087
fits_file_b = fits.open('/Users/erin/Documents/Projects/KOA_Climate/data/KOA_104087/NIRSPEC/extracted/fitstbl/flux/NS.20000325.42919_47_flux_tbl.fits.gz')

data_class_b = fits_file_b[1] # Get the [1]th element of fits_file
data_b = data_class_b.data # Get the data from data_class

b = data_b['wave (A)'] # Get the wavelength column
r = data_b['flux (cnts)'] # Get the flux column
b = b / 10000 # Convert the wavelength from Angstroms to microns
```

```python
# KOA 109549
fits_file_c = fits.open('/Users/erin/Documents/Projects/KOA_Climate/data/KOA_109549/NIRSPEC/extracted/fitstbl/flux/NS.20000325.45382_47_flux_tbl.fits.gz')

data_class_c = fits_file_c[1] # Get the [1]th element of fits_file
data_c = data_class_c.data # Get the data from data_class

c = data_c['wave (A)'] # Get the wavelength column
s = data_c['flux (cnts)'] # Get the flux column
c = c / 10000 # Convert the wavelength from Angstroms to microns
```

```python
# KOA_115150
fits_file_d = fits.open('/Users/erin/Documents/Projects/KOA_Climate/data/KOA_115150/NIRSPEC/extracted/fitstbl/flux/NS.20000325.45429_47_flux_tbl.fits.gz')

data_class_d = fits_file_d[1] # Get the [1]th element of fits_file
data_d = data_class_d.data # Get the data from data_class

d = data_d['wave (A)'] # Get the wavelength column
t = data_d['flux (cnts)'] # Get the flux column
d = d / 10000 # Convert the wavelength from Angstroms to microns
```

```python
# Subplot test 
fig, axs = plt.subplots(2,2, figsize = (20,10))
fig.tight_layout() 

plt.subplots_adjust(left = 0.125, bottom = 0.1, right = 0.9, top = 0.9, wspace = 0.2, hspace = 0.2)

axs[0,0].plot(a,q, color = 'k')
axs[0,0].set_title('KOA_462')

axs[0,1].plot(b,r, color = 'r')
axs[0,1].set_title('KOA_104087')

axs[1,0].plot(c,s, color = 'b')
axs[1,0].set_title('KOA_109549')

axs[1,1].plot(d,t, color = 'm')
axs[1,1].set_title('KOA_115150')

for ax in axs.flat:
    ax.set(xlabel = 'Wavelength ($\mu$m)', ylabel = 'Flux (counts)')

```
