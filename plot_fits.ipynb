```python
fits_file = fits.open('/Users/erin/Documents/Projects/KOA_Climate/data/KOA_462/NIRSPEC/extracted/fitstbl/flux/NS.20000325.45349_47_flux_tbl.fits.gz')
fits_file.info() # Looking at the fits data
```

```python
data_class = fits_file[1] # Get the [1]th element of fits_file
data = data_class.data # Get the data from data_class
print(data.columns) # Show the data
```

```python
a = data['wave (A)'] # Get the wavelength column
q = data['flux (cnts)'] # Get the flux column
a = a / 10000 # Convert the wavelength from Angstroms to microns
```

```python
fs = 40 # Fontsize
ts = 26 # Tick size
plt.figure(figsize = (20,10)) # Figure size
plt.plot(a, q, color = 'k', label = 'data') # plotting wavelength vs. flux and coloring the data

plt.xlim([1.605, 1.625]) # Limits of the x-axis
plt.xticks(fontsize = ts) # Tick size on the x-axis
plt.yticks(fontsize = ts) # Tick size on the y-axis

plt.xlabel('Wavelength ($\mu$m)', fontsize = fs) # Labelling the x-axis and applying fontsize
plt.ylabel('Flux (counts)', fontsize = fs) # Labelling the y-axis and applying fontsize
plt.title('Plot #1', fontsize = fs) # Giving the plot a title and applying fontsize

plt.show() # Show the plot
```
